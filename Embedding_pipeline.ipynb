```python
#STEP 2 — Run the Embedding Script 
import pandas as pd
from sentence_transformers import SentenceTransformer

df = pd.read_csv('reviews.csv')
texts = df['text'].tolist()

model = SentenceTransformer('all-MiniLM-L6-v2')

batch_size = 32

embeddings = model.encode(texts, batch_size=batch_size, show_progress_bar=True)

import numpy as np
np.save('review_embeddings.npy', embeddings)
```

```python
#STEP 3 — Inspect the Embeddings 
import numpy as np

emb = np.load('review_embeddings.npy')
print("Embedding shape:", emb.shape)
print("First vector (first 10 dims):", emb[0][:10])
```

```python
#STEP 4 — Generate a t‑SNE Plot 
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import numpy as np

embeddings = np.load("review_embeddings.npy") 

tsne = TSNE(n_components=2, perplexity=3, max_iter=300)

tsne_results = tsne.fit_transform(embeddings)

plt.figure(figsize=(10, 6))
plt.scatter(tsne_results[:, 0], tsne_results[:, 1])
plt.title('t-SNE visualization of product embeddings')
plt.xlabel('t-SNE feature 1')
plt.ylabel('t-SNE feature 2')
plt.show()
```
